```python
print('ETAPE3: comfy + proxy token-gate')
import os, threading, subprocess, sys, time, asyncio, signal
from aiohttp import web, ClientSession
TOKEN = os.environ.get('HEPHAISTOS_TOKEN') or '__KAGGLE_COMFY_TOKEN__'
try:
    from kaggle_secrets import UserSecretsClient
    TOKEN = UserSecretsClient().get_secret('HEPHAISTOS_TOKEN') or TOKEN
    print('token via kaggle secrets')
except Exception as e:
    print('token via env/placeholder:', type(e).__name__)
comfy = subprocess.Popen([sys.executable,'main.py','--listen','127.0.0.1','--port','8188','--disable-cuda-malloc'],
    cwd='/kaggle/working/ComfyUI')
async def proxy(req):
    if req.headers.get('X-Hephaistos-Token') != TOKEN:
        return web.Response(status=401, text='bad token')
    async with ClientSession() as s:
        url = 'http://127.0.0.1:8188' + req.path_qs
        body = await req.read() if req.can_read_body else None
        async with s.request(req.method, url, data=body,
            headers={k:v for k,v in req.headers.items() if k.lower() not in ('host','x-hephaistos-token')}) as up:
            return web.Response(status=up.status, body=await up.read(),
                headers={k:v for k,v in up.headers.items() if k.lower() not in ('content-encoding','content-length','transfer-encoding')})
async def shutdown(req):
    if req.headers.get('X-Hephaistos-Token') != TOKEN:
        return web.Response(status=401, text='bad token')
    threading.Timer(2.0, lambda: os.kill(os.getpid(), signal.SIGTERM)).start()
    return web.Response(text='extinction dans 2s')
async def main():
    app = web.Application()
    app.router.add_route('*','/shutdown',shutdown)
    app.router.add_route('*','/{path:.*}',proxy)
    r = web.AppRunner(app); await r.setup()
    await web.TCPSite(r,'0.0.0.0',8080).start()
    print('proxy pret :8080')
threading.Thread(target=lambda: asyncio.run(main()), daemon=True).start()
MAX_H = float(os.environ.get('KERNEL_MAX_HOURS','8'))
threading.Timer(MAX_H*3600, lambda: os.kill(os.getpid(), signal.SIGTERM)).start()
print('auto-extinction dans', MAX_H, 'h')
time.sleep(20)
print('comfy poll:', comfy.poll(), '(None = vivant)')
print('ETAPE3 FINIE')
# + SD 1.5 pruned (demande utilisateur)
dst15 = os.path.join(d,'v1-5-pruned-emaonly.safetensors')
if not os.path.exists(dst15):
    r = subprocess.run(['wget','-q','--show-progress','-O',dst15,
      'https://huggingface.co/Comfy-Org/stable-diffusion-v1-5-archive/resolve/main/v1-5-pruned-emaonly_fp16.safetensors'],capture_output=True,text=True)
    print('wget v1-5 rc=', r.returncode)
print('v1-5 size:', os.path.getsize(dst15) if os.path.exists(dst15) else 'ABSENT')

# + RealVisXL V5 fp16 (photorealisme, ~7Go)
d2 = os.path.join(d,'RealVisXL_V5.0_fp16.safetensors')
if not os.path.exists(d2):
    r = subprocess.run(['wget','-q','--show-progress','-O',d2,
      'https://huggingface.co/SG161222/RealVisXL_V5.0/resolve/main/RealVisXL_V5.0_fp16.safetensors'],capture_output=True,text=True)
    print('wget realvis rc=', r.returncode)
print('realvis size:', os.path.getsize(d2) if os.path.exists(d2) else 'ABSENT')

# + DreamShaperXL Turbo v2 (rapide, ~6.5Go)
d3 = os.path.join(d,'DreamShaperXL_Turbo_v2.safetensors')
if not os.path.exists(d3):
    r = subprocess.run(['wget','-q','--show-progress','-O',d3,
      'https://huggingface.co/Lykon/dreamshaper-xl-v2-turbo/resolve/main/DreamShaperXL_Turbo_v2.safetensors'],capture_output=True,text=True)
    print('wget dsxl rc=', r.returncode)
print('dsxl size:', os.path.getsize(d3) if os.path.exists(d3) else 'ABSENT')
```

```python
print('ETAPE4: tunnel')
import subprocess, re, time, threading, os
if not os.path.exists('/kaggle/working/cloudflared'):
    r = subprocess.run(['wget','-q','https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64','-O','/kaggle/working/cloudflared'],capture_output=True,text=True)
    print('cloudflared dl rc=', r.returncode)
    subprocess.run(['chmod','+x','/kaggle/working/cloudflared'],check=False)
    _cf_cmd = (['/kaggle/working/cloudflared','tunnel','--no-autoupdate','run','--token','__MAROTTE_TUNNEL_TOKEN__'] if '__MAROTTE_TUNNEL_TOKEN__' else ['/kaggle/working/cloudflared','tunnel','--no-autoupdate','--url','http://127.0.0.1:8188'])
proc = subprocess.Popen(_cf_cmd,
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
url = [None]
def pump():
    for line in proc.stdout:
        print('CF:', line.strip()[:160])
        m = re.search('https://[a-z0-9-]+\.trycloudflare\.com', line)
        if m and not url[0]: url[0] = m.group(0)
threading.Thread(target=pump, daemon=True).start()
for _ in range(40):
    if url[0]: break
    time.sleep(2)
print('>>> KAGGLE_COMFY_URL =', url[0])
```

```python
# WATCHDOG + KEEP-ALIVE : relance cloudflared ET comfy s'ils meurent
# -> surveille ce log : a chaque (re)demarrage du tunnel la NOUVELLE URL s'affiche
import time, subprocess, os, signal, re, threading
cf = [None]
def start_cf():
    _cf_cmd = (['/kaggle/working/cloudflared','tunnel','--no-autoupdate','run','--token','__MAROTTE_TUNNEL_TOKEN__'] if '__MAROTTE_TUNNEL_TOKEN__' else ['/kaggle/working/cloudflared','tunnel','--no-autoupdate','--url','http://127.0.0.1:8188'])
    cf[0] = subprocess.Popen(_cf_cmd,
        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    def pump():
        for line in cf[0].stdout:
            m = re.search('https://[a-z0-9-]+\.trycloudflare\.com', line)
            if m:
                print('>>> TUNNEL ACTIF:', m.group(0), flush=True)
            elif ' ERR ' in line or 'error' in line.lower():
                print('CF-ERR:', line.strip()[:140], flush=True)
    threading.Thread(target=pump, daemon=True).start()
def start_comfy():
    global comfy
    comfy = subprocess.Popen([sys.executable,'main.py','--listen','127.0.0.1','--port','8188','--disable-cuda-malloc'],
        cwd='/kaggle/working/ComfyUI')
print('watchdog actif - /shutdown ou kaggle kernels stop pour arreter', flush=True)
while True:
    if cf[0] is None or cf[0].poll() is not None:
        print('WATCHDOG: cloudflared mort -> redemarrage', flush=True)
        start_cf()
    if comfy.poll() is not None:
        print('WATCHDOG: comfy mort -> redemarrage', flush=True)
        start_comfy()
    time.sleep(20)
```
